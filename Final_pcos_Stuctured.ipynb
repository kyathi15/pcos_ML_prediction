```python
# ============================================================
# PCOS MACHINE LEARNING PROJECT
# COMPLETE WORK DONE SO FAR
# DATASET → UNDERSTANDING → CLEANING → EDA
# → PREPROCESSING → TRAIN/TEST → MODELS → EVALUATION
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    ConfusionMatrixDisplay,
    roc_auc_score,
    roc_curve
)

print("Libraries imported successfully!")


# ============================================================
# 2. UPLOAD DATASET
# ============================================================

uploaded = files.upload()

filename = next(iter(uploaded))

print("\nUploaded file:", filename)


# ============================================================
# 3. LOAD DATASET
# ============================================================

df = pd.read_excel(filename)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())


# ============================================================
# 4. BASIC DATA UNDERSTANDING
# ============================================================

print("\n" + "=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print("\nNumber of rows and columns:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())


# ============================================================
# 5. TARGET VARIABLE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("TARGET VARIABLE")
print("=" * 60)

print("\nPCOS target distribution:")
print(df["PCOS (Y/N)"].value_counts())

print("\nPCOS target percentage:")
print(
    df["PCOS (Y/N)"].value_counts(normalize=True) * 100
)


# ============================================================
# 6. REMOVE IDENTIFIER / SOURCE COLUMNS
# ============================================================

df = df.drop(
    columns=[
        "Sl. No",
        "Patient File No.",
        "Data_Source"
    ],
    errors="ignore"
)

print("\nIdentifier/source columns removed successfully!")

print("New dataset shape:", df.shape)


# ============================================================
# 7. CHECK DUPLICATES AFTER CLEANING
# ============================================================

print("\nDuplicate rows after cleaning:")
print(df.duplicated().sum())


# ============================================================
# 8. SEPARATE FEATURES AND TARGET
# ============================================================

X = df.drop(
    "PCOS (Y/N)",
    axis=1
)

y = df["PCOS (Y/N)"].map({
    "N": 0,
    "Y": 1
})

print("\n" + "=" * 60)
print("FEATURES AND TARGET")
print("=" * 60)

print("\nX shape:")
print(X.shape)

print("\ny shape:")
print(y.shape)

print("\nTarget distribution:")
print(y.value_counts())


# ============================================================
# 9. CATEGORICAL AND NUMERICAL FEATURES
# ============================================================

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("\n" + "=" * 60)
print("FEATURE TYPES")
print("=" * 60)

print("\nCategorical features:")
print(categorical_features)

print(
    "\nNumber of categorical features:",
    len(categorical_features)
)

print(
    "\nNumber of numerical features:",
    len(numerical_features)
)


# ============================================================
# 10. PCOS DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    x=y.map({
        0: "No PCOS",
        1: "PCOS"
    })
)

plt.title("PCOS Distribution")
plt.xlabel("PCOS Status")
plt.ylabel("Number of Patients")
plt.show()


# ============================================================
# 11. AGE DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.histplot(
    data=df,
    x="Age (yrs)",
    kde=True
)

plt.title("Age Distribution")
plt.xlabel("Age (years)")
plt.ylabel("Number of Patients")
plt.show()


# ============================================================
# 12. BMI DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.histplot(
    data=df,
    x="BMI",
    kde=True
)

plt.title("BMI Distribution")
plt.xlabel("BMI")
plt.ylabel("Number of Patients")
plt.show()


# ============================================================
# 13. BMI VS PCOS
# ============================================================

plot_df = df.copy()

plot_df["PCOS Status"] = plot_df[
    "PCOS (Y/N)"
].map({
    "N": "No PCOS",
    "Y": "PCOS"
})

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=plot_df,
    x="PCOS Status",
    y="BMI"
)

plt.title("BMI vs PCOS Status")
plt.xlabel("PCOS Status")
plt.ylabel("BMI")
plt.show()


# ============================================================
# 14. AGE VS PCOS
# ============================================================

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=plot_df,
    x="PCOS Status",
    y="Age (yrs)"
)

plt.title("Age vs PCOS Status")
plt.xlabel("PCOS Status")
plt.ylabel("Age")
plt.show()


# ============================================================
# 15. SYMPTOM ANALYSIS
# ============================================================

symptom_columns = [
    "Weight gain(Y/N)",
    "hair growth(Y/N)",
    "Skin darkening (Y/N)",
    "Hair loss(Y/N)",
    "Pimples(Y/N)"
]

for col in symptom_columns:

    plt.figure(figsize=(7, 5))

    sns.countplot(
        data=plot_df,
        x=col,
        hue="PCOS Status"
    )

    plt.title(f"{col} vs PCOS Status")
    plt.xlabel(col)
    plt.ylabel("Number of Patients")

    plt.show()


# ============================================================
# 16. HORMONE ANALYSIS
# ============================================================

hormone_columns = [
    "FSH(mIU/mL)",
    "LH(mIU/mL)",
    "AMH(ng/mL)",
    "PRL(ng/mL)",
    "Vit D3 (ng/mL)"
]

for col in hormone_columns:

    plt.figure(figsize=(7, 5))

    sns.boxplot(
        data=plot_df,
        x="PCOS Status",
        y=col
    )

    plt.title(f"{col} vs PCOS Status")
    plt.xlabel("PCOS Status")
    plt.ylabel(col)

    plt.show()


# ============================================================
# 17. NUMERICAL SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("NUMERICAL SUMMARY")
print("=" * 60)

display(df.describe())


# ============================================================
# 18. CORRELATION HEATMAP
# ============================================================

numeric_df = df.select_dtypes(
    include=["int64", "float64"]
)

plt.figure(figsize=(18, 12))

sns.heatmap(
    numeric_df.corr(),
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")

plt.show()


# ============================================================
# 19. OUTLIER ANALYSIS USING IQR
# ============================================================

outlier_summary = []

for col in numerical_features:

    Q1 = X[col].quantile(0.25)

    Q3 = X[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR

    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (X[col] < lower_bound) |
        (X[col] > upper_bound)
    ).sum()

    outlier_summary.append({
        "Feature": col,
        "Outlier Count": outlier_count
    })


outlier_df = pd.DataFrame(
    outlier_summary
)

outlier_df = outlier_df.sort_values(
    by="Outlier Count",
    ascending=False
)

print("\n" + "=" * 60)
print("OUTLIER ANALYSIS")
print("=" * 60)

display(outlier_df)


# ============================================================
# 20. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n" + "=" * 60)
print("TRAIN-TEST SPLIT")
print("=" * 60)

print("\nTraining data:", X_train.shape)

print("Testing data:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())


# ============================================================
# 21. CREATE PREPROCESSOR
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[

        (
            "num",
            StandardScaler(),
            numerical_features
        ),

        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        )
    ]
)

print("\nPreprocessor created successfully!")


# ============================================================
# 22. PREPROCESS DATA
# ============================================================

X_train_processed = preprocessor.fit_transform(
    X_train
)

X_test_processed = preprocessor.transform(
    X_test
)

print(
    "\nProcessed training data shape:",
    X_train_processed.shape
)

print(
    "Processed testing data shape:",
    X_test_processed.shape
)


# ============================================================
# 23. CREATE MACHINE LEARNING MODELS
# ============================================================

models = {

    "Logistic Regression":
        LogisticRegression(
            max_iter=1000
        ),

    "Decision Tree":
        DecisionTreeClassifier(
            random_state=42
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=200,
            random_state=42
        ),

    "SVM":
        SVC(
            probability=True,
            random_state=42
        ),

    "Gradient Boosting":
        GradientBoostingClassifier(
            random_state=42
        )
}

print("\nModels created successfully!")


# ============================================================
# 24. TRAIN ALL MODELS
# ============================================================

trained_models = {}

for name, model in models.items():

    print(f"Training {name}...")

    model.fit(
        X_train_processed,
        y_train
    )

    trained_models[name] = model

print("\nAll models trained successfully!")


# ============================================================
# 25. GENERATE PREDICTIONS
# ============================================================

predictions = {}

for name, model in trained_models.items():

    predictions[name] = model.predict(
        X_test_processed
    )

print("\nPredictions generated successfully!")


# ============================================================
# 26. MODEL COMPARISON
# ============================================================

comparison = []

for name, model in trained_models.items():

    y_pred = model.predict(
        X_test_processed
    )

    comparison.append({

        "Model": name,

        "Accuracy":
            accuracy_score(
                y_test,
                y_pred
            ),

        "Precision":
            precision_score(
                y_test,
                y_pred,
                zero_division=0
            ),

        "Recall":
            recall_score(
                y_test,
                y_pred,
                zero_division=0
            ),

        "F1 Score":
            f1_score(
                y_test,
                y_pred,
                zero_division=0
            )
    })


comparison_df = pd.DataFrame(
    comparison
)

print("\n" + "=" * 60)
print("MODEL COMPARISON")
print("=" * 60)

display(comparison_df)


# ============================================================
# 27. CONFUSION MATRICES
# ============================================================

for name, model in trained_models.items():

    y_pred = model.predict(
        X_test_processed
    )

    ConfusionMatrixDisplay.from_predictions(
        y_test,
        y_pred,
        display_labels=[
            "No PCOS",
            "PCOS"
        ]
    )

    plt.title(
        f"Confusion Matrix - {name}"
    )

    plt.show()


# ============================================================
# 28. ROC-AUC
# ============================================================

auc_results = []

for name, model in trained_models.items():

    y_prob = model.predict_proba(
        X_test_processed
    )[:, 1]

    auc = roc_auc_score(
        y_test,
        y_prob
    )

    auc_results.append({

        "Model": name,

        "ROC-AUC": auc
    })


auc_df = pd.DataFrame(
    auc_results
)

print("\n" + "=" * 60)
print("ROC-AUC COMPARISON")
print("=" * 60)

display(auc_df)


# ============================================================
# 29. ROC CURVES
# ============================================================

plt.figure(figsize=(8, 6))

for name, model in trained_models.items():

    y_prob = model.predict_proba(
        X_test_processed
    )[:, 1]

    fpr, tpr, _ = roc_curve(
        y_test,
        y_prob
    )

    auc = roc_auc_score(
        y_test,
        y_prob
    )

    plt.plot(
        fpr,
        tpr,
        label=f"{name} (AUC = {auc:.3f})"
    )


plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")

plt.ylabel("True Positive Rate")

plt.title("ROC Curves")

plt.legend()

plt.show()


# ============================================================
# 30. CLASSIFICATION REPORTS
# ============================================================

for name, model in trained_models.items():

    y_pred = model.predict(
        X_test_processed
    )

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print(
        classification_report(
            y_test,
            y_pred,
            labels=[0, 1],
            target_names=[
                "No PCOS",
                "PCOS"
            ],
            zero_division=0
        )
    )


# ============================================================
# 31. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("PCOS MACHINE LEARNING PROJECT")
print("COMPLETED UP TO MODEL EVALUATION")
print("=" * 70)

print("\nDataset records:", len(df))

print("Number of features:", X.shape[1])

print("Training records:", len(X_train))

print("Testing records:", len(X_test))

print("\nModels trained:")

for name in trained_models.keys():
    print("-", name)

print("\nModel comparison:")
display(comparison_df)

print("\nROC-AUC comparison:")
display(auc_df)

print("\nEverything completed successfully!")
```

```python
# ============================================================
# PCOS ML PROJECT
# NEXT STAGE:
# MODEL SELECTION → HYPERPARAMETER TUNING
# → FINAL MODEL → SAVE MODEL → PREDICTION
# ============================================================


# ============================================================
# 32. SELECT MODEL BASED ON F1 SCORE
# ============================================================

print("=" * 70)
print("MODEL SELECTION")
print("=" * 70)

# Find the model with the highest F1 Score
best_model_name = comparison_df.loc[
    comparison_df["F1 Score"].idxmax(),
    "Model"
]

best_f1_score = comparison_df["F1 Score"].max()

print("\nSelected model based on F1 Score:")
print(best_model_name)

print("\nF1 Score:", round(best_f1_score, 4))


# ============================================================
# 33. DISPLAY ALL MODEL RESULTS
# ============================================================

print("\nModel Performance Comparison:")
display(
    comparison_df.sort_values(
        by="F1 Score",
        ascending=False
    )
)


# ============================================================
# 34. HYPERPARAMETER TUNING
# ============================================================

from sklearn.model_selection import GridSearchCV


# ------------------------------------------------------------
# Choose parameter grid according to selected model
# ------------------------------------------------------------

if best_model_name == "Logistic Regression":

    param_grid = {
        "C": [0.01, 0.1, 1, 10, 100],
        "solver": ["liblinear", "lbfgs"]
    }

    tuning_model = LogisticRegression(
        max_iter=2000
    )


elif best_model_name == "Decision Tree":

    param_grid = {
        "max_depth": [3, 5, 10, None],
        "min_samples_split": [2, 5, 10],
        "min_samples_leaf": [1, 2, 4]
    }

    tuning_model = DecisionTreeClassifier(
        random_state=42
    )


elif best_model_name == "Random Forest":

    param_grid = {
        "n_estimators": [100, 200],
        "max_depth": [None, 10, 20],
        "min_samples_split": [2, 5],
        "min_samples_leaf": [1, 2]
    }

    tuning_model = RandomForestClassifier(
        random_state=42
    )


elif best_model_name == "SVM":

    param_grid = {
        "C": [0.1, 1, 10],
        "kernel": ["linear", "rbf"],
        "gamma": ["scale", "auto"]
    }

    tuning_model = SVC(
        probability=True,
        random_state=42
    )


elif best_model_name == "Gradient Boosting":

    param_grid = {
        "n_estimators": [100, 200],
        "learning_rate": [0.05, 0.1],
        "max_depth": [2, 3, 5]
    }

    tuning_model = GradientBoostingClassifier(
        random_state=42
    )


# ============================================================
# 35. RUN GRID SEARCH
# ============================================================

print("\nStarting hyperparameter tuning...")
print("Selected model:", best_model_name)

grid_search = GridSearchCV(
    estimator=tuning_model,
    param_grid=param_grid,
    cv=5,
    scoring="f1",
    n_jobs=-1,
    verbose=1
)

grid_search.fit(
    X_train_processed,
    y_train
)

print("\nHyperparameter tuning completed!")


# ============================================================
# 36. BEST PARAMETERS
# ============================================================

print("\n" + "=" * 70)
print("BEST HYPERPARAMETERS")
print("=" * 70)

print("\nBest parameters:")
print(grid_search.best_params_)

print("\nBest cross-validation F1 Score:")
print(round(grid_search.best_score_, 4))


# ============================================================
# 37. FINAL TUNED MODEL
# ============================================================

final_model = grid_search.best_estimator_

print("\nFinal tuned model:")
print(final_model)


# ============================================================
# 38. FINAL MODEL PREDICTIONS
# ============================================================

final_predictions = final_model.predict(
    X_test_processed
)

print("\nFinal predictions generated successfully!")


# ============================================================
# 39. FINAL MODEL EVALUATION
# ============================================================

final_accuracy = accuracy_score(
    y_test,
    final_predictions
)

final_precision = precision_score(
    y_test,
    final_predictions,
    zero_division=0
)

final_recall = recall_score(
    y_test,
    final_predictions,
    zero_division=0
)

final_f1 = f1_score(
    y_test,
    final_predictions,
    zero_division=0
)

final_probabilities = final_model.predict_proba(
    X_test_processed
)[:, 1]

final_auc = roc_auc_score(
    y_test,
    final_probabilities
)


print("\n" + "=" * 70)
print("FINAL MODEL PERFORMANCE")
print("=" * 70)

print("\nModel:", best_model_name)

print("Accuracy :", round(final_accuracy, 4))
print("Precision:", round(final_precision, 4))
print("Recall   :", round(final_recall, 4))
print("F1 Score :", round(final_f1, 4))
print("ROC-AUC  :", round(final_auc, 4))


# ============================================================
# 40. FINAL CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("FINAL CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        final_predictions,
        labels=[0, 1],
        target_names=[
            "No PCOS",
            "PCOS"
        ],
        zero_division=0
    )
)


# ============================================================
# 41. FINAL CONFUSION MATRIX
# ============================================================

plt.figure(figsize=(6, 5))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    final_predictions,
    display_labels=[
        "No PCOS",
        "PCOS"
    ]
)

plt.title(
    f"Final Model Confusion Matrix - {best_model_name}"
)

plt.show()


# ============================================================
# 42. FINAL ROC CURVE
# ============================================================

fpr, tpr, thresholds = roc_curve(
    y_test,
    final_probabilities
)

plt.figure(figsize=(7, 5))

plt.plot(
    fpr,
    tpr,
    label=f"AUC = {final_auc:.3f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title(
    f"ROC Curve - Final Model ({best_model_name})"
)

plt.legend()

plt.show()


# ============================================================
# 43. SAVE FINAL MODEL + PREPROCESSOR
# ============================================================

import joblib

joblib.dump(
    final_model,
    "final_pcos_model.pkl"
)

joblib.dump(
    preprocessor,
    "pcos_preprocessor.pkl"
)

print("\nFinal model saved as:")
print("final_pcos_model.pkl")

print("\nPreprocessor saved as:")
print("pcos_preprocessor.pkl")


# ============================================================
# 44. DOWNLOAD MODEL FILES
# ============================================================

from google.colab import files

files.download(
    "final_pcos_model.pkl"
)

files.download(
    "pcos_preprocessor.pkl"
)


# ============================================================
# 45. CREATE A PREDICTION FUNCTION
# ============================================================

def predict_pcos(input_data):

    # Convert input into DataFrame
    input_df = pd.DataFrame(
        [input_data]
    )

    # Apply the same preprocessing
    processed_input = preprocessor.transform(
        input_df
    )

    # Predict
    prediction = final_model.predict(
        processed_input
    )[0]

    # Probability
    probability = final_model.predict_proba(
        processed_input
    )[0][1]

    if prediction == 1:
        result = "PCOS"
    else:
        result = "No PCOS"

    return result, probability


# ============================================================
# 46. TEST PREDICTION USING AN EXISTING PATIENT
# ============================================================

# Take one patient from the test dataset
sample_patient = X_test.iloc[0]

sample_result, sample_probability = predict_pcos(
    sample_patient.to_dict()
)

print("\n" + "=" * 70)
print("SAMPLE PCOS PREDICTION")
print("=" * 70)

print("\nPrediction:", sample_result)

print(
    "PCOS probability:",
    round(sample_probability * 100, 2),
    "%"
)


# ============================================================
# 47. PROJECT FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("PCOS MACHINE LEARNING PROJECT")
print("FINAL SUMMARY")
print("=" * 70)

print("\nDataset records:", len(df))

print("Features used:", X.shape[1])

print("Training records:", len(X_train))

print("Testing records:", len(X_test))

print("\nModels evaluated:")
for name in trained_models.keys():
    print("-", name)

print("\nSelected model:", best_model_name)

print(
    "Final Accuracy:",
    round(final_accuracy, 4)
)

print(
    "Final Precision:",
    round(final_precision, 4)
)

print(
    "Final Recall:",
    round(final_recall, 4)
)

print(
    "Final F1 Score:",
    round(final_f1, 4)
)

print(
    "Final ROC-AUC:",
    round(final_auc, 4)
)

print("\nProject completed successfully! 🎉")
```
